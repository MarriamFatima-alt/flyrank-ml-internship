## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Decision:** which pages enter this week's editorial review queue, out of far more candidates than there is time to check by hand.

**Who acts:** a content/SEO editor working through a fixed-size weekly queue.

**Cost of a wrong call:** missing a genuine decliner lets it keep losing traffic silently until someone happens to notice — a slow, compounding cost. A false alarm costs a few minutes of an editor's time on a page that turns out to be fine. Because those costs are asymmetric, the scoring should rank generously rather than filter aggressively — it's a queue, not a gate.

```python
print("Unit of analysis : one content page (content_id)")
print("Acts on it       : content/SEO editor, weekly review queue")
print("Wrong-call costs : missed decliner = slow silent traffic loss")
print("                   false alarm     = a few minutes of review time")
```
